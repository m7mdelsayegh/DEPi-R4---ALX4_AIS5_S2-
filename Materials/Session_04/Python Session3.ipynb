```python
# Task1
# Create gussing number game

# secrete num
# while -> inoput  == secrete num = you win break
# else try again
```

```python
import random
secret_num = random.randint(1,100)
tries = 0

while True:
    input_num = int(input("Enter a number: "))
    tries += 1
   
    if input_num == secret_num:
        print(f"{secret_num} is the secret number You win after {tries} tries")
        break
    elif input_num > secret_num:
        print("Too high")
    else:
        print("Too low")
```

```python
def pow (num):
    return num * num
```

```python
resullt = map(pow , number)  # for loop   -> brodcasting  element -> map
print(list(resullt))
```
