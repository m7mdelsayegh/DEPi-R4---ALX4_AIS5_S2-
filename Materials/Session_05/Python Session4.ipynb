```python
import time

class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def show_balance(self):
        print(f"The available balance of {self.name} is {self.balance:.2f} EGP")      


    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            time.sleep(2)
            print("Successful transaction")   

        else:
            raise ValueError ("Amount must be greater than 0")       
        

    def withdraw(self, amount):
        if amount > 0 and  amount < self.balance:
            self.balance -= amount
            time.sleep(2)
            print("Successful transaction")

        else: 
            raise ValueError ("Amount must be greater than 0 and less than the available balance")    
        

    def __str__ (self):
        return f"Name: {self.name}\nBalance: {self.balance:.2f} EGP"
    
    def __repr__(self):
        return (f"BankAccount({self.name}, {self.balance})")
    

    def __add__ (self, other): 
        return self.balance + other.balance
```
