```python
# Task1 
# Create a shoping cart
# create 2 lists one for products and one for prices
# use while loop to accept data from user
# if product ==q break
# after this find the total amount of the cart
```

```python
prodcuts = []
prices = []
while True:
    product = input("Please enter product name: ")
    if product == "q":
        break
    price = int(input("Please enter product price: "))
    prodcuts.append(product)
    prices.append(price)
print(f"Total amount of the cart is {sum(prices)}")
```

```python
cars = {
    "car1":2500,
    "car2":3000,
    "car3":4000
}


totla_amount = 0
while True:
    product = input("Please enter product name: ")
    if product == "q":
        break
    elif product in cars.keys():
        totla_amount += cars[product]
                
print(f"Total amount of the cart is {totla_amount}")
```
